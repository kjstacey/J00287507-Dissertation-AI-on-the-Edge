## 6. Local Account Authentication and Physical Access

**Investigator-reported configuration** (no separate screenshot of account or passcode settings is included in this branch):

- The designated headset remained signed into the researcher's personal Meta account.
- A local headset unlock passcode was **not** configured.
- Normal headset use did not require fresh account authentication or local credential entry.
- The experimental application was installed only on this Quest 3S.

**Interpretation:** An individual with physical possession of the accessible headset might interact with functions exposed to the signed-in user. This does **not** show that account passwords, protected application data, or unrestricted developer access would be obtainable. No unauthorized physical-access attempt was made.

## 7. Developer Mode and USB Debugging Authorization

**Investigator-reported development workflow:**

1. Enable Developer Mode on the experimental Quest 3S.
2. Connect the headset to the development workstation by USB.
3. Approve the headset's **USB debugging** authorization prompt on each connection.
4. Use Meta Quest Developer Hub to inspect the connected headset.
5. Build and deploy the experimental application using Unity **Build and Run**.

Per-connection approval was reported; **persistent “always allow this computer” authorization was not established**. Developer Mode and a debugging interface do not, on their own, demonstrate a bypass of the approval prompt.

**Potential concern:** Once a connection is authorized, development interfaces provide capabilities for package deployment and inspection. Misuse by an authorized or compromised development workstation is a relevant scenario, but was **not tested** here.

## 8. Unity APK Deployment and Packaged Assets

**Investigator-reported configuration and source-review findings from implementation materials kept in separate branches:**

- Unity's **Build and Run** installed the experimental APK through the authorized development connection.
- The ONNX model and controlled benchmark files were included as project assets in the application build.
- The reviewed local inference code loaded a project-assigned Unity `ModelAsset` using `ModelLoader.Load(...)`.
- Benchmark inputs were provided through packaged `TextAsset` references where applicable.
- Local inference did not require a separate runtime download of the model or benchmark dataset.

**Security boundary:** APK packaging is **not equivalent to encryption or cryptographic authenticity verification**. No separate model-hash, signature-verification, or secure update mechanism was identified in the reviewed inference scripts. This is a statement about the reviewed application path, not a complete audit of the Android package or operating-system signature checks.

**Not tested:** extracting assets from the APK; repackaging an APK; modifying ONNX weights; inspecting on-device application sandbox permissions; malware installation; bypassing Android package signatures.

**Related branches:** local Quest Unity code and associated performance/data notebooks contain the actual implementation and benchmark methodology; their contents are not duplicated here.

## 9. Passthrough Camera Permission and Data Path

**Investigator-reported permission:** the experimental application was **explicitly granted camera access** to use the Quest passthrough camera during development and testing.

**Source-review observations:**

- The Quest camera script referenced `PassthroughCameraAccess`.
- It checked the camera object and acquisition state before processing frames.
- Its image path accessed the camera texture/colors for local image preprocessing and inference.

The permission grant was a required application access control. It did **not** establish the authenticity of the objects or visual scene captured by the camera. A misleading scene could affect inference integrity without implying a camera-permission bypass.

**Evidence limitation:** Neither of the two screenshots is a dedicated per-application camera-permission screenshot. The permission state is documented from the investigator's report and the functional application workflow, not from an exported headset permission screen.

**Not tested:** denied-permission behavior, permissions of unrelated applications, camera stream interception, or adversarial physical input manipulation.

## 11. Contextual Research — Quest 2 Ransomware

**PDF:** [`Tahat et al. - 2024 - VR Headset Ransomware Attack Vulnerability.pdf`](Tahat%20et%20al.%20-%202024%20-%20VR%20Headset%20Ransomware%20Attack%20Vulnerability.pdf)

**Citation:** M. Z. Tahat, W. B. Glisson, and B. Al Smadi, “VR Headset Ransomware Attack Vulnerability,” *2024 IEEE Canadian Conference on Electrical and Computer Engineering (CCECE)*, DOI: [10.1109/CCECE59415.2024.10667339](https://doi.org/10.1109/CCECE59415.2024.10667339). Zotero key: `tahat_vr_2024`.

Key findings and relevance:

- **Study target:** Oculus Quest 2, an earlier device/software environment, **not** the experimental Quest 3S.
- **Procedure:** enabled Developer Mode, authorized USB debugging/file access, and sideloaded test APKs with SideQuest (PDF p. 3).
- **Results:** seven of ten tested Android ransomware families produced successful effects, including device lockout or data encryption (PDF p. 4, Table III).
- **Additional test:** a purpose-built ransomware proof of concept was also executed on the Quest 2 (PDF pp. 4–5).

**Applicability to this research:** illustrates consequences of installing **untrusted APKs through developer workflows**. It provides no direct evidence that a Quest 3S on the documented Horizon OS configuration is vulnerable to these same samples. The dissertation's application was built and deployed by the investigator; **no ransomware was installed or executed**.

## 14. Scope and Limitations

- Qualitative inspection only; **no penetration testing, malware deployment, or exploit reproduction**.
- The automatic update on November 5 prevents assigning the observed firmware build to all preceding experiments.
- No app-permission screenshot, original APK binary, APK signature report, package scan output, or on-device filesystem inspection accompanies this branch.
- No evidence of a headset unlock passcode or other local authentication step in the reported experimental workflow; account-level protections were not independently audited.
- A development workstation had USB debugging privileges **after** headset authorization. No unauthorized debugging access was demonstrated.
- Package-integrity and camera input-integrity concerns were not tested experimentally.
- Quest 2 ransomware results were treated as **contextual research**, not as measured behavior of the Quest 3S.
- The local-inference review does not establish security properties of cloud request handling, encryption in transit, cloud authentication, or Google Cloud Run; those belong to the cloud-assisted security subsection.

Reproducing this notebook means **repeating the configuration review and evidence interpretation**, not reproducing a security exploit.

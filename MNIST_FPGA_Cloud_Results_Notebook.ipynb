1. **Formal cloud measurements** using controlled 100-image MNIST batches for accuracy, latency, throughput, and communication checks.

## 12. Memory interpretation

The reduced cloud configuration still required PS-side memory for networking and TLS.

Approximate major application buffers:

```text
TLS receive buffer        32 KB
Mbed TLS memory pool     128 KB
HTTP response buffer       4 KB
HTTP header buffer       ~512 B
784-byte image buffer
```

Additional memory was required by lwIP, DHCP, DNS, TCP state, Mbed TLS structures, stack, heap, and application state.

This memory footprint therefore represents the cloud communication/application path rather than local neural-network model storage.

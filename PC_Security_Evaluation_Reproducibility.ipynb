# PC Security Evaluation — HP ZBook Fury 17 G7

This notebook documents the qualitative security evaluation of the **local Windows PC inference environment** used in the dissertation experiments.

The purpose is to preserve:

- the workstation and development-software configuration;
- Windows security and account-access evidence collected on **November 5, 2026**;
- local model, benchmark data, and archival storage conditions;
- Unity source-code and package-manifest review;
- published Unity security advisory applicability;
- Lumens VC-B2U USB camera and Windows camera-permission observations; and
- the evidence boundaries and limits of the security review.

The review supports **Chapter IV, Section 4.7.2.2 — PC Deployment**. Cloud request handling, TLS, and Google Cloud Run security are treated separately in the cloud-assisted inference evaluation.

This notebook records **how the configuration was examined and what evidence supported the findings**. It does not document penetration testing or demonstrated exploitation.

## 2. Security Evaluation Sequence

```text
1. Record the hardware and Windows version
2. Examine Windows Security status and Defender results
3. Examine firewall profiles and installed updates
4. Inspect UEFI, Secure Boot, TPM, VBS, and DMA settings
5. Review account privileges and Windows Hello sign-in
6. Inspect BitLocker states for internal and attached storage
7. Record project/model/dataset storage locations and access conditions
8. Review Unity version, package manifest, and representative C# scripts
9. Compare Unity security advisories with the actual version and code paths
10. Examine the Lumens USB camera and Windows camera-access history
11. Record findings, applicability, and limitations
```

The approach was **configuration inspection, source review, and vendor advisory matching**. No adversarial tests were conducted.

## 13. Model Loading and Controlled Benchmark Inputs

The reviewed local model loader and representative inference scripts used:

```csharp
runtimeModel = ModelLoader.Load(modelAsset);
worker = new Worker(runtimeModel, BackendType.GPUCompute);
```

The local PC camera example used `BackendType.CPU`; CPU and GPU Compute execution were both part of the PC benchmark environment.

The representative accuracy test used Unity-assigned `ModelAsset` and `TextAsset` references, and checked the expected sizes of the MNIST image and label binaries before processing them:

```text
10,000 images × 784 floats × 4 bytes per float
10,000 expected labels
```

These size checks confirmed the expected input layout; they did **not** prove data authenticity or rule out maliciously altered values of the correct length.

No runtime model download, dynamic user-selected model path, or Brotli decoding call was identified in the reviewed local scripts. This conclusion is limited to those scripts, not the entire Editor or every package dependency.

**Evidence:** `camera_access_settings.png`

The Lumens **VC-B2U** camera was connected **directly over USB**, not as a network camera. The investigator confirmed that **when connected, Lumens was the default camera**.

In `CameraMNISTInference.cs`, the application used:

The script read local image pixels, applied grayscale conversion and thresholding, formed 28 × 28 input tensors, and performed local inference. It did not transmit images to a server.

**Finding:** The selected first-enumerated camera matched the actual Lumens default configuration. No explicit identity verification appeared in the reviewed script, so substitution or enumeration changes were potential input-integrity concerns, not observed incidents.

## 19. Reproduction Sequence

To reproduce the **qualitative review procedure** for a comparably configured workstation:

```text
[ ] Record exact Windows release, build, workstation, and CPU
[ ] Capture Windows Security overview and threat-protection status
[ ] Record firewall profile states and active profile
[ ] Record update history and the review date
[ ] Inspect UEFI, Secure Boot, TPM, VBS, and Kernel DMA status
[ ] Record account privileges and Windows sign-in method
[ ] Check drive-encryption status for each relevant volume
[ ] Document research-file location and actual access permissions
[ ] Record Unity Editor and Inference Engine versions
[ ] Review manifest and representative model/data loading scripts
[ ] Compare official Unity advisories with versions and execution paths
[ ] Record camera connection, enumeration, and Windows access settings
[ ] Separate confirmed configuration from possible attacks or unverified states
```

These steps reproduce the **inspection method**, not the original Windows configuration or a penetration test. The original screenshots and recorded observations remain the evidence for this dissertation's implementation.

## 20. Evaluation Limitations

This review was based on screenshots, source-file examination, published documentation, and investigator-reported conditions. It did **not** include:

- penetration testing or local privilege-escalation attempts;
- comprehensive Defender or third-party malware scanning;
- filesystem ACL exports or forensic drive analysis;
- cryptographic verification of the ONNX model or binary datasets;
- independent validation of all installed Windows updates and trust-store contents;
- detailed inspection of Unity package-manager network activity or all dependencies;
- deliberate camera substitution or spoofing;
- vulnerability exploitation, memory attacks, or system compromise; or
- measurements of security-related performance overhead.

**Not observed**, **not implemented in reviewed code**, **not independently verified**, and **not exploitable** are not interchangeable conclusions.

All screenshots were collected during the November 2026 review. Earlier settings might have differed. The external research archive on `D:` was created after the experimental runs, so it must not be presented as the active data location during testing.

# MNIST Unity Local Quest 3S Benchmarking

This notebook documents the benchmarking workflow and supporting evidence for the **local Meta Quest 3S MNIST implementation**.

It begins after the Quest application has already been configured, built, installed, and validated as described in the separate setup notebook.

The purpose of this notebook is to document:

- the benchmark sequence;
- GPU and CPU backend testing;
- tethered and untethered execution conditions;
- OVR Metrics / MQDH logging;
- controlled MNIST accuracy, latency, and throughput tests;
- sustained inference;
- frozen-camera inference;
- end-to-end camera processing;
- live camera classification and throughput; and
- the screenshot and log files used as evidence.

This notebook records the experimental procedure and evidence. Cross-platform interpretation and dissertation-level discussion are handled separately.

## 1. Experimental Platform

The Quest benchmarks were performed on a **Meta Quest 3S** using the Unity application documented in the Quest setup notebook.

The project used:

- Unity Editor 6000.3.22f1;
- Meta XR SDK All-in-One 207.0.1;
- Unity Inference Engine 2.6.1;
- Unity OpenXR 1.16.1;
- the shared MNIST ONNX model;
- the same 28×28 preprocessing logic used in the PC/FPGA comparison; and
- Meta Quest Developer Hub (MQDH) / OVR Metrics tools for device-side observations.

The local Quest application does not require cloud inference for these tests.

## 2. Benchmark Execution Order

The benchmark evidence was collected in the following overall sequence:

```text
1. GPU — tethered
2. GPU — untethered
3. CPU — tethered
4. CPU — untethered
```

The screenshots and logs were preserved in the order in which the test sessions were run.

This sequence is important when interpreting repeated screenshots of the same benchmark.

## 3. Tethered and Untethered Screenshot Rule

For the screenshots, repeated captures of the same benchmark are interpreted using the following rule:

- when multiple screenshots exist for the same test, the **first half** correspond to the tethered run;
- the **second half** correspond to the untethered run; and
- when only one screenshot exists for a test, it is treated as **untethered**.
This rule is used throughout this notebook when identifying the evidence files.

## 5. Backend Configuration

The Quest GPU configuration uses:

```csharp
BackendType.GPUCompute
```

For CPU testing, the local Unity worker was switched to:

```csharp
BackendType.CPU
```

The MNIST model and benchmark procedures remained the same. Output labels were adjusted as needed so that screenshots and logs identified the selected backend.

The purpose of the CPU/GPU comparison is therefore to compare execution backends on the same Quest device rather than different trained models.

## 6. Benchmark Categories

The Quest measurements fall into five groups:

```text
Controlled MNIST tests
    → accuracy
    → latency
    → throughput

Sustained inference
    → long-running repeated inference

Frozen-camera benchmark
    → repeated inference on a fixed camera-derived 28×28 tensor

End-to-end camera benchmark
    → camera acquisition + preprocessing + tensor creation + inference + readback

Live camera behavior
    → classification
    → unique-frame throughput
```

OVR Metrics logs provide device-side context during these runs.

## 7. Controlled MNIST Accuracy, Latency, and Throughput

The controlled benchmark scripts use the shared MNIST test binaries rather than the live camera.

The test set contains:

```text
10,000 images
784 float values per image
10,000 expected labels
```

The model accuracy target is the same validated result used for the local PC implementation:

```text
9663 / 10000
96.63%
```

Latency and throughput use the same general benchmark logic as the other Unity platforms so that the results remain comparable.

## 12. Frozen-Camera Benchmark Method

The frozen-camera benchmark uses the current camera-derived 28×28 image as a fixed test input.

The benchmark uses:

```text
50 warm-up runs
1000 measured runs
```

For each measured run, the existing tensor is scheduled for inference and the output is synchronously read back.

The reported statistics include:

- mean;
- median;
- P95;
- minimum;
- maximum; and
- standard deviation.

Because preprocessing has already occurred, this benchmark isolates repeated inference on a camera-derived input.

## 15. End-to-End Camera Benchmark Method

The end-to-end camera benchmark includes the full local camera processing path:

```text
Quest camera image
    ↓
camera pixel access
    ↓
mapping to FPGA-equivalent 160×120 geometry
    ↓
28×28 sampling
    ↓
grayscale conversion
    ↓
binary threshold
    ↓
tensor creation
    ↓
local inference
    ↓
synchronous output readback
```

The benchmark uses:

```text
50 warm-up runs
1000 measured runs
```

This measurement therefore includes substantially more work than the frozen-camera benchmark.

## 19. Live Camera Throughput Method

Live throughput differs from the controlled 10,000-image throughput benchmark.

The camera script processes **new camera frames** and counts completed predictions over elapsed real time.

The test therefore reflects application-level camera behavior rather than the maximum rate at which a fixed data source can be fed to the model.

## 29. Reporting Guidance

The screenshots serve as visual evidence of benchmark output, while the OVR Metrics logs provide device-level telemetry.

For final dissertation tables:

- use the Unity benchmark output for accuracy, latency, and throughput values;
- use the camera benchmark output for frozen and end-to-end camera timing;
- use the live-throughput output for application-level camera throughput;
- use the OVR Metrics logs for device resource/performance context; and
- keep tethered and untethered conditions distinct when reporting GPU results.

Repeated screenshots should be preserved when they represent separate runs or different connection conditions.
